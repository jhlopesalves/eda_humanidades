### 1.1 Leis de De Morgan

As **Leis de De Morgan** descrevem duas equivalências entre negação, conjunção
e disjunção. Para duas condições Booleanas $A$ e $B$:

$$
\neg(A \land B) \equiv (\neg A) \lor (\neg B)
$$

$$
\neg(A \lor B) \equiv (\neg A) \land (\neg B)
$$

Em Python, as mesmas relações podem ser escritas como:

```python
not (A and B)
(not A) or (not B)
```
e
```python
not (A or B)
(not A) and (not B)
```
As duas expressões imprimem o mesmo valor - e imprimirão o mesmo valor para
qualquer `nacionalidade`: é a segunda lei de De Morgan em funcionamento.

````python
materia = "História"
nivel = "A"
idade = 16
nacionalidade = "Brasileiro"

condicao = (
    materia == "História"
    and nivel == "A"
    and idade == 16
    and (
        nacionalidade == "Brasileiro"
        or nacionalidade == "Argentino"
    )
)

print(condicao)
````

````python
nacionalidade_valida = (
    nacionalidade == "Brasileiro"
    or nacionalidade == "Argentino"
)

print(not nacionalidade_valida)

print(
    nacionalidade != "Brasileiro"
    and nacionalidade != "Argentino"
)
````

### 1.1 Tokenização

**Tokenização** é a segmentação de uma sequência de texto em unidades menores,
chamadas **tokens**. A definição operacional de um token depende das regras
utilizadas pelo tokenizador.

Neste exemplo, `split()` sem argumentos realiza uma tokenização simplificada:
a string é separada em regiões delimitadas por espaços em branco. Sequências
consecutivas de espaços, quebras de linha ou tabulações funcionam como
separadores.

Essa operação não separa automaticamente a pontuação. Assim, `words,` e
`words` seriam tokens diferentes neste procedimento.

A unidade contada também muda. `len(sample_text)` conta caracteres;
`len(tokens)` contará os elementos produzidos pela tokenização.

````python
tokens = sample_text.split()

print(tokens)
print(len(tokens))
````

### 1.2 Indexação e slicing

Strings e listas são sequências ordenadas. Cada elemento ocupa uma posição
identificada por um **índice**, começando em `0`.

Um único índice seleciona um elemento:

`sequencia[indice]`

Um **slice** seleciona uma subsequência:

`sequencia[inicio:fim]`

O elemento em `inicio` é incluído; o elemento em `fim` não é incluído.

A forma completa é:

`sequencia[inicio:fim:passo]`

`inicio`, `fim` e `passo` podem ser omitidos quando os valores padrão são
adequados. Índices negativos contam posições a partir do final da sequência.

````python
sample_text[10:100]
````

````python
print(tokens[0:3])

print(tokens[:3])

print(tokens[3:])

print(tokens[::4])

print(tokens[-2])

print(tokens[::-1])
````

## 3. Iteração com `for`

Um loop `for` permite executar um bloco de código repetidamente, uma vez para
cada elemento produzido por um objeto **iterável**. Strings e listas são
exemplos de objetos iteráveis já utilizados neste curso.

A estrutura básica é:

```python
for item in iteravel:
    # bloco executado para cada item
```

A cada iteração, `item` recebe o próximo elemento do iterável. O nome dessa
variável é arbitrário e deve, quando possível, indicar o tipo de elemento que
está sendo percorrido.

A indentação delimita o bloco pertencente ao loop. Todas as instruções
indentadas sob o `for` são executadas para cada elemento.

````python
tokens = "stats is everywhere in research".split()

for token in tokens:
    print(token)
````

### 3.1 Estado durante a iteração

Um loop também pode atualizar valores definidos antes de sua execução. Um
contador, por exemplo, pode registrar quantas iterações já foram realizadas.

O contador começa em `0` e é incrementado em uma unidade a cada passagem pelo
bloco do `for`.

````python
numero_iteracao = 0

for token in tokens:
    numero_iteracao += 1
    print(f"{numero_iteracao}: {token}")
````

## 4. Controle de fluxo: `if`, `elif` e `else`

Estruturas condicionais permitem executar blocos diferentes de código de acordo
com o resultado de expressões Booleanas.

Uma estrutura condicional pode conter três tipos de cláusula:

- `if` introduz a primeira condição;
- `elif` introduz condições adicionais, avaliadas apenas se as anteriores forem
  falsas;
- `else` define o bloco executado quando nenhuma das condições anteriores é
  verdadeira.

Python avalia as condições na ordem em que aparecem. Em uma cadeia formada por
`if`, `elif` e `else`, apenas o primeiro bloco cuja condição seja verdadeira é
executado.

Assim como em um loop `for`, a indentação delimita as instruções pertencentes a
cada bloco.

No exemplo abaixo, uma lista vazia recebe os tokens cujo comprimento excede
cinco caracteres. O método `.append()` acrescenta um elemento ao final da
lista.

````python
texto = (
"""
A statistical hypothesis test is a method of statistical inference used to decide whether the data provide sufficient evidence to reject a particular hypothesis. A statistical hypothesis test typically involves a calculation of a test statistic. Then a decision is made, either by comparing the test statistic to a critical value or equivalently by evaluating a p-value computed from the test statistic. Roughly 100 specialized statistical tests are in use and noteworthy.
"""
)

tokens = texto.split()
tokens_longos = []

for token in tokens:
    if len(token) > 5:
        tokens_longos.append(token)

print(tokens_longos)
````

### 4.1 Condições mutuamente exclusivas

Uma cadeia com `if`, `elif` e `else` pode classificar cada elemento em uma única
categoria.

No exemplo seguinte, o comprimento de cada token determina uma categoria
didática. Os limites escolhidos servem apenas para demonstrar o controle de
fluxo.

````python
for token in tokens:
    comprimento = len(token)

    if comprimento <= 3:
        categoria = "curto"
    elif comprimento <= 6:
        categoria = "médio"
    else:
        categoria = "longo"

    print(f"{token}: {categoria}")
````

### 5.1 Media aritmética

Para $n$ valores numéricos $x_1, x_2, \ldots, x_n$, a média aritmética é
definida por:

$$
\bar{x}
=
\frac{1}{n}\sum_{i=1}^{n}x_i
=
\frac{x_1 + x_2 + \cdots + x_n}{n}
$$

O cálculo requer duas quantidades: a soma dos valores e o número de valores
observados. Em Python, essas operações podem ser obtidas diretamente com
`sum()` e `len()`.

````python
valores = [10, 20, 30, 40, 50]

n = len(valores)
soma = sum(valores)
média = soma / n

print(f"n = {n}")
print(f"soma = {soma}")
print(f"média = {média}")
````

## 6. Dicionários

Um **dicionário** (`dict`) associa **chaves** a **valores**. Cada chave
identifica um valor dentro da estrutura:

`chave: valor`

As chaves de um dicionário são únicas e precisam ser objetos **hashable**.
Para os exemplos introdutórios deste curso, strings e números são escolhas
comuns para chaves.

Os valores não precisam ter o mesmo tipo e podem representar diferentes tipos
de informação.

Dicionários são particularmente úteis quando queremos recuperar um valor a
partir de um identificador, em vez de sua posição na sequência.

````python
frequencias = {
    "verbo": 25,
    "substantivo": 10,
    "artigo": 29,
}

print(frequencias)
````

### 6.1 Chaves e acesso aos valores

Em um dicionário, o operador `in` verifica a presença de uma **chave**.

O teste:

`"verbo" in frequencias`

portanto, verifica se `"verbo"` é uma das chaves do dicionário; ele não procura
esse valor entre as frequências armazenadas.

````python
print("verbo" in frequencias)
print("substantivo" in frequencias)
````

### 6.2 Acesso por chave e `KeyError`

Um valor pode ser recuperado diretamente por sua chave utilizando colchetes:

`dicionario[chave]`

Se a chave não existir, Python interrompe a operação e produz um `KeyError`.

A célula seguinte contém deliberadamente uma chave ausente. A mensagem de erro
faz parte do exemplo e deve ser lida antes de continuar.### Acessando chaves e lendo um `KeyError`

Execute esta célula e leia a mensagem de erro antes de continuar.

````python
lista_de_tags = ["verbo", "substantivo", "preposição", "artigo"]

frequencias = {
    "verbo": 25,
    "substantivo": 10,
    "artigo": 29,
}

for tag in lista_de_tags:
    print(f"tag: {tag}: {frequencias[tag]}")
````

````python
lista_tags = ["verbo", "substantivo", "preposição", "artigo"]

frequencias = {
    "verbo": 37,
    "substantivo": 15,
    "preposição": 29,
}

for tag in lista_tags:
    print(f"{tag}: {frequencias.get(tag, 0)}")
````

O `KeyError` ocorreu porque `"preposição"` não é uma chave de `frequencias`.
Como a exceção interrompeu o loop, os elementos posteriores não foram
processados.

Quando a ausência de uma chave é uma possibilidade esperada, o método `.get()`
permite consultá-la sem produzir um `KeyError`.

A forma geral é:

`dicionario.get(chave, valor_padrao)`

Se a chave existir, seu valor é retornado. Caso contrário, `.get()` retorna o
valor padrão especificado.

````python
for tag in lista_de_tags:
    valor = frequencias.get(tag, "Valor não encontrado")
    print(f"{tag}: {valor}")
````

### 6.3 Valores padrão e significado dos dados

O valor padrão utilizado em `.get()` deve ser compatível com o significado dos
dados.

Em um dicionário de **contagens**, `0` pode ser apropriado quando a ausência de
uma chave significa que nenhuma ocorrência daquela categoria foi registrada.

Essa escolha não é geral. Uma chave ausente também pode representar informação
desconhecida ou não disponível; nesse caso, substituí-la automaticamente por
zero alteraria o significado dos dados.

````python
frequencias = {
    "verbo": 37,
    "substantivo": 15,
    "preposição": 29,
}

for tag in lista_de_tags:
    frequencia = frequencias.get(tag, 0)
    print(f"{tag}: {frequencia}")
````
