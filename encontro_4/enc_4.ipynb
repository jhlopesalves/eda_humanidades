## 2. De listas para `np.arrays`

Até aqui, utilizamos **listas** e **dicionários** para armazenar sequências de valores. Essas estruturas são bastante flexíveis e atendem bem a tarefas gerais, mas apresentam limitações importantes quando o objetivo é realizar **cálculos numéricos em larga escala**.

Para superar essas limitações, a biblioteca **NumPy** introduz uma nova estrutura de dados: o **`array`** (ou *ndarray*, *n-dimensional array*). Trata-se de um "container" **homogêneo**; todos os elementos têm o mesmo tipo; o que permite executar operações numéricas de maneira muito mais rápida e econômica do que com listas.

A partir dessa estrutura surge o conceito central do encontro: a **vetorização**.

> **Vetorização** é a prática de aplicar uma operação a um conjunto inteiro de valores de uma só vez, sem precisar escrever um loop explícito sobre cada elemento.

Em vez de pensarmos *"para cada elemento da sequência, faça X"*, passamos a pensar *"faça X sobre toda a sequência"*.

````python
lista_de_numeros = [4, 7, 3, 9, 5]

array_de_numeros = np.array(lista_de_numeros)

print(lista_de_numeros)
display(array_de_numeros)
print(type(array_de_numeros))
````

````python
# Em uma lista comum, esta operação não soma 1 a cada elemento.
print(lista_de_numeros * 2)

# Em um array NumPy, a operação é aplicada elemento por elemento.
display(array_de_numeros * 2)
display(array_de_numeros + 1)
display(array_de_numeros > 5)
````

A diferença importante:

- em listas Python, `lista * 2` repete a lista;
- em arrays NumPy, `array * 2` multiplica cada valor por 2.

````python
print(np.mean(array_de_numeros))
print(np.median(array_de_numeros))
print(np.min(array_de_numeros))
print(np.max(array_de_numeros))
````

## 4. Uma lista de dicionários

Um dicionário representa uma obra.

Uma lista de dicionários representa várias obras.

Essa estrutura é muito útil porque cada dicionário continua agrupando as informações de uma única linha.

Em outras palavras:
```
- cada dicionário será uma linha;
- cada chave será uma coluna;
- cada valor será uma célula.
```

````python
obras = [
    {
        "autor": "John Tukey",
        "titulo": "Exploratory Data Analysis",
        "area": "estatística",
        "ano": 1977,
        "n_tokens": 125_000,
        "n_types": 35_900,
    },
    {
        "autor": "Robert Merton",
        "titulo": "The Normative Structure of Science",
        "area": "sociologia",
        "ano": 1942,
        "n_tokens": 7_600,
        "n_types": 2_650,
    },
    {
        "autor": "Karl Popper",
        "titulo": "The Logic of Scientific Discovery",
        "area": "filosofia",
        "ano": 1934,
        "n_tokens": 155_000,
        "n_types": 12_800,
    },
]
obras
````

## 5. Transformando registros em tabela

Agora transformaremos nossa lista de dicionários em um `DataFrame`.

Um `DataFrame` é uma tabela em pandas.

````python
obras_df = pd.DataFrame(data=obras)

obras_df
````

## 8. Criando uma coluna derivada: TTR

Podemos criar novas colunas a partir de colunas existentes.

Aqui vamos calcular a TTR:

$$
\text{TTR} = \frac{\text{número de types}}{\text{número de tokens}}
$$


````python
obras_df["ttr"] = obras_df["n_types"] / obras_df["n_tokens"]

obras_df[["titulo", "n_tokens", "n_types", "ttr"]]
````

````python
n_sentencas = [5_200, 415, 5_950, 3_500]

obras_df["n_sentencas"] = n_sentencas

obras_df
````

````python
obras_df["tokens_por_sentenca"] = (
    obras_df["n_tokens"] / obras_df["n_sentencas"]
)

obras_df[["titulo", "n_tokens", "n_sentencas", "tokens_por_sentenca"]]
````

## 9.  Criando categorias com `np.where()`

O `np.where()` cria valores diferentes dependendo de uma condição.

A estrutura é:
`np.where(condição, valor_se_true, valor_se_falso)`

````python
obras_df["classe_tamanho"] = np.where(
    obras_df["n_tokens"] >= 100_000, # condição
    "longo", # se a condição for verdadeira
    "curto", # se a condição for falsa
)

obras_df
````

## 13. Agrupando dados

Agrupar significa separar os dados por categorias e calcular uma métrica dentro de cada grupo.

A pergunta geral é:

> como uma medida muda de um grupo para outro?

A estrutura básica é:

```python
df.groupby("grupo")["coluna_numerica"].metrica()

Leia da direita pra esquerda:
> A média do número de tokens agrupado pela 'classe_tamanho'

````python
obras_df.groupby("classe_tamanho")["n_tokens"].mean()
````

````python
# Utilizando "named aggregations" (agregações nomeadas)
# A estrutura é: nome_da_nova_coluna=("coluna_original", "metrica")
obras_df.groupby("classe_tamanho").agg(
    # Conta a quantidade de obras (títulos) em cada grupo
    n_obras=("titulo", "count"),

    # Calcula a média da coluna 'n_tokens' para cada grupo
    media_tokens=("n_tokens", "mean"),

    # Calcula a média da coluna 'ttr' para cada grupo
    media_ttr=("ttr", "mean"),
)
````
