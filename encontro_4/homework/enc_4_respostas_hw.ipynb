#### 7. Ordene os países em 2007 por expectativa de vida

#### 8. Para a coluna `lifeExp` (expectativa de vida), calcule separadamente: **média**, **mediana**, **mínimo** e **máximo**.

**Critério**: Guarde cada resultado numa variável e imprima cada métrica usando f-strings.

```python
media_lifeExp = gapminder_df["lifeExp"].mean()
mediana_lifeExp = gapminder_df["lifeExp"].median()
minimo_lifeExp = gapminder_df["lifeExp"].min()
maximo_lifeExp = gapminder_df["lifeExp"].max()

print(f"Média: {media_lifeExp:.2f}")
print(f"Mediana: {mediana_lifeExp:.2f}")
print(f"Mínimo: {minimo_lifeExp:.2f}")
print(f"Máximo: {maximo_lifeExp:.2f}")
```

#### 13. Crie uma nova coluna categórica chamada `country_size` segundo a regra:

- população **> 100 milhões** -> `"large"`
- caso contrário -> `"small"`

Utilize uma função em **NumPy** que aplica uma condição elemento a elemento e retorna um valor para o caso `True` e outro para o `False`.



> Psssssst: utilize underlines ( `_` ) para digitar números grandes em Python; por exemplo, `1 milhão == 1_000_000`.

```python
gapminder_df["country_size"] = np.where(gapminder_df["pop"] > 100_000_000, "large", "small")
gapminder_df.head()
```
