### Sobre o dataset `Gapminder`

O Gapminder é uma fundação sueca independente fundada por Hans Rosling, Ola Rosling e Anna Rosling Rönnlund, dedicada a combater equívocos generalizados sobre o desenvolvimento global por meio de dados públicos e visualizações acessíveis. A organização compila e publica indicadores socioeconômicos provenientes de fontes primárias como o `Banco Mundial`, a `ONU`, a `OMS` e institutos nacionais de estatística.

O arquivo utilizado neste notebook é o recorte clássico popularizado pelo pacote `gapminder` do R (Jennifer Bryan, 2015).

**Descrição das colunas:**

| Coluna | Descrição |
|---|---|
| `country` | Nome do país (codificação ASCII, em inglês). |
| `continent` | Continente de referência (5 categorias: Africa, Americas, Asia, Europe, Oceania). |
| `year` | Ano de referência da observação (1952–2007, passo de 5 anos). |

**Notas metodológicas relevantes:**

- O ajuste por `PPP (Purchasing Power Parity)` torna `gdpPercap` comparável entre países e ao longo do tempo, ao corrigir diferenças de custo de vida e inflação. Isto é o que permite comparar, digamos, o PIB per capita do Vietnam em 1972 com o da Noruega em 2007 numa mesma escala
