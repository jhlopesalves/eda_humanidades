# Entendendo o output:

### 1. Tokens e tipos: a distinção fundamental!

Antes de qualquer métrica, dois conceitos:

- **Token** == cada *ocorrência* de palavra no texto. Se "autor" aparece 50 vezes, são 50 tokens.
- **Type** (ou *tipo*) == cada palavra distinta. "Casa" aparecendo 50 vezes conta como **1 type**.

A TTR é uma razão entre types e tokens; a razão hapax também é uma proporção. Já o Top 10 apresenta itens e contagens, não razões.

## 2. TTR - Type-Token Ratio (Razão Tipo/Token)

**O que mede:** a proporção de types entre os tokens observados sob esta tokenização e este pré-processamento.

**Fórmula:**

$$
\text{TTR} = \frac{\text{types}}{\text{tokens}} = \frac{V}{N}
$$

```python
print(f"TTR: {metricas['ttr']:.4f}")
```

**Cuidado importante:**

> `TTR` **depende fortemente e de forma mecânica do tamanho do texto**. Textos longos tendem a apresentar TTR menor, mesmo sem uma diferença substantiva de variedade lexical.

Por isso, a comparação direta requer textos de extensões semelhantes e o mesmo procedimento de tokenização e pré-processamento. Quando os tamanhos diferem, medidas como `MATTR` ou `MTLD` podem ser mais adequadas. Não há uma faixa universal de TTR válida para todas as línguas, gêneros e decisões de pré-processamento.

## 3. Hapax legomena

**Etimologia:** do grego ἅπαξ λεγόμενον - "*dito uma só vez*". O termo vem da filologia clássica, usado por estudiosos de Homero para marcar palavras que aparecem uma única vez no corpus homérico.

**Definição:** palavra cuja frequência no texto é exatamente **1**.

**Nosso exemplo:**

```python
print(f"Hapax legomena: {metricas['hapax']:,}")
print(f"Razão hapax: {metricas['razao_hapax']:.4f}")
```

$$
\frac{\text{hapax}}{\text{types}}
$$

Distribuições de frequência lexical frequentemente apresentam uma cauda longa, mas não existe uma faixa universal para a razão hapax. O valor depende do tamanho e da composição do texto ou corpus, da língua e de sua morfologia, do gênero e registro, da tokenização e do uso ou não de lematização.

**Por que isso é interessante?**

- **Linguisticamente:** muitos hapax são nomes próprios, neologismos, termos técnicos ou palavras usadas num contexto muito específico. Eles carregam alta carga informacional.
- **Filologicamente:** hapax servem como pistas para datação de textos, atribuição de autoria e identificação de interpolações.
- **Metodologicamente:** em corpora grandes, hapax também são bons objetos de inspeção, porque podem revelar erros de processamento, problemas de OCR, typos, tokenização inadequada ou palavras mal normalizadas. Nem todo hapax é um achado linguístico: alguns são sinais de que o pipeline de limpeza precisa ser revisado.

### Observação estatística: a Lei de Zipf

A **Lei de Zipf** prevê que, ordenando as palavras por frequência (*rank* 1 = mais frequente), a frequência decai aproximadamente como:

$$
f(r) \approx \frac{C}{r^{s}}, \quad s \approx 1
$$

Esse modelo descreve uma queda aproximadamente inversa entre *rank* e frequência. Os dados reais não precisam seguir exatamente metade, um terço e assim por diante; o ajuste é uma regularidade aproximada a ser verificada.
